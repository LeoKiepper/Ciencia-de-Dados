# Questão 1

Mostre que a distância de *Manhattan* (vide Aula 8) é uma métrica.

A distância de Manhattan pode ser definida como segue:
Dados $\mathbf{x} \in \mathbb{R}^m$ e $\mathbf{y} \in \mathbb{R}^m$:

$$d(\mathbf{x}, \mathbf{y}) =  \Vert \mathbf{x} - \mathbf{y} \Vert_1  = {\sum_{i=1}^m |x_i - y_i|}. $$

Uma função de dissmilaridade é dita uma métrica quando possui as seguintes propriedades:
1. $d(x, y) \ge 0$
2. $d(x, y) = 0 \iff x = y$
3. $d(x, y) = d(y, x)$
4. $d(x, y) \le d(x, z) + d(z, y)$.

### **Demonstração**
1. A distância de Manhattan é definida como uma soma de módulos. Por definição, $|x| \ge 0$, e como a soma de reais positivos sempre resulta num real positivo, então $d(\mathbf{x}, \mathbf{y}) \ge 0$ 

    $\quad\square$

2. 
    - $d(x, y) = 0 \Rightarrow x = y$:$\quad$ $\sum_{i=1}^m |x_i - y_i| > 0$ ocorrerá sempre que $\mathbf{x} \ne \mathbf{y}$, pois, nessa condição, os componentes de $\mathbf{x}$ vão diferir de $\mathbf{y}$ em pelo menos uma coordenada, e o somando será não nulo em cada coordenada em que isto ocorrer. Mas, como demonstrado em 1, $d(\mathbf{x}, \mathbf{y}) \ge 0$. Assim, por contraposição, $d(\mathbf{x},\mathbf{y}) = 0 \Rightarrow \mathbf{x} = \mathbf{y}$. 

    - $x = y \Rightarrow d(x, y) = 0$:$\quad$ Se $\mathbf{x} = \mathbf{y}$, as coordenadas de $\mathbf{x}$ e $\mathbf{y}$ serão iguais em todas as dimensões, então $\sum_{i=1}^m |x_i - y_i| = 0$

    $\quad\square$

3. Dentro do somando $|x_i - y_i|$, $x_i$ e $y_i$ são reais, então, pela simetria da operação de módulo $|x| = |-x|$ sobre os reais, pode-se afirmar que $|x_i - y_i| = |y_i - x_i|$. Portanto, $d(\mathbf{x}, \mathbf{y}) =  {\sum_{i=1}^m |x_i - y_i|} = {\sum_{i=1}^m |y_i - x_i|} = d(\mathbf{y}, \mathbf{x})$

    $\quad\square$

4. $d(x, z) + d(z, y) \stackrel{\text{def}}{=} \sum_{i=1}^m |x_i - z_i| + \sum_{i=1}^m |z_i - y_i| \quad= \sum_{i=1}^m |x_i - z_i| + |z_i - y_i|$. Pela desigualdade triangular, $|u+v| \le |u| + |v|$, e faça $u = x_i - z_i$ e $v = z_i - y_i$. Logo: $|(x_i - z_i) + (z_i - y_i)| = |x_i - y_i| \;\;\le\;\; |x_i - z_i| + |z_i - y_i|$. Como a soma preserva desigualdades, pode-se fazer $\sum_{i=1}^m |x_i - y_i| \le \sum_{i=1}^m |x_i - z_i| + |z_i - y_i|$. Isto é, $d(\mathbf{x}, \mathbf{y}) \le  d(\mathbf{x}, \mathbf{z}) + d(\mathbf{z}, \mathbf{y})$. 

    $\quad\square$


# Questão 2

Mostre que a distância euclidiana (vide Aula 8) é uma métrica.

**Dica:** você pode usar a desigualdade de *Cauchy-Schwarz*.

A distância Euclidian é definida como segue:
Dados $\mathbf{x} \in \mathbb{R}^m$ e $\mathbf{y} \in \mathbb{R}^m$:

$$d(\mathbf{x}, \mathbf{y}) =  \Vert \mathbf{x} - \mathbf{y} \Vert_2  = \sqrt{\sum_{i=1}^m (x_i - y_i)^2}.$$

Uma função de dissmilaridade é dita uma métrica quando possui as seguintes propriedades:
1. $d(x, y) \ge 0$
2. $d(x, y) = 0 \iff x = y$
3. $d(x, y) = d(y, x)$
4. $d(x, y) \le d(x, z) + d(z, y)$.

### **Demonstração**
1. O contradomínio da função raíz quadrada é $\left\{ y \in \mathbb{R} \,|\, y \ge 0 \right\}$.

    $\quad\square$

2. 
    - $d(x, y) = 0 \Rightarrow x = y$:$\quad$ $\sum_{i=1}^m (x_i - y_i)^2 > 0$ ocorrerá sempre que $\mathbf{x} \ne \mathbf{y}$, pois, nessa condição, os componentes de $\mathbf{x}$ vão diferir de $\mathbf{y}$ em pelo menos uma coordenada, e o somando será não nulo em cada coordenada em que isto ocorrer. Mas, como demonstrado em 1, $d(\mathbf{x}, \mathbf{y}) \ge 0$. Assim, por contraposição, $d(\mathbf{x},\mathbf{y}) = 0 \Rightarrow \mathbf{x} = \mathbf{y}$. 

    - $x = y \Rightarrow d(x, y) = 0$:$\quad$ Se $\mathbf{x} = \mathbf{y}$, as coordenadas de $\mathbf{x}$ e $\mathbf{y}$ serão iguais em todas as dimensões, então $\sum_{i=1}^m (x_i - y_i) = 0$

    $\quad\square$

3. Dentro do somando $(x_i - y_i)^2$, $x_i$ e $y_i$ são reais, então, pela simetria da operação $x^2 = (-x)^2$ sobre os reais, pode-se afirmar que $(x_i - y_i) = (y_i - x_i)$. Portanto, $d(\mathbf{x}, \mathbf{y}) =  \sqrt{\sum_{i=1}^m (x_i - y_i)^2} = \sqrt{\sum_{i=1}^m (y_i - x_i)^2} = d(\mathbf{y}, \mathbf{x})$

    $\quad\square$

4. Sejam os pontos em questão posicionados como no diagrama abaixo:
![Triangulo](https://raw.githubusercontent.com/LeoKiepper/Ciencia-de-Dados/main/lista02/triangulo.svg)

Seja o vetor $\mathbf{w} = \mathbf{u} + \mathbf{v}$, cuja norma (ao quadrado) pode ser escrita por $\Vert \mathbf{w} \Vert ^ 2 = \langle \mathbf{w},\mathbf{w} \rangle = \langle \mathbf{u} + \mathbf{v} , \mathbf{u} + \mathbf{v} \rangle = \langle \mathbf{u} , \mathbf{u} \rangle + 2 \langle \mathbf{u} , \mathbf{v} \rangle + \langle \mathbf{v} , \mathbf{v} \rangle$. Pela desigualdade de Cauchy-Schwarz: $|\langle \mathbf{u},\mathbf{v} \rangle| \le \Vert \mathbf{u} \Vert\cdot\Vert \mathbf{v} \Vert$. Quando aplicado aos reais, sempre se pode afirmar sobre a operação de módulo que $x \le |x|$, então $x$ pode ser usado no lugar de $|x|$.
Logo, escreve-se $\langle \mathbf{u},\mathbf{v} \rangle \le \Vert\mathbf{u} \Vert\cdot\Vert \mathbf{v} \Vert$. Perturbando essa expressão, obtém-se $\Vert \mathbf{u} \Vert^2 + 2\langle \mathbf{u},\mathbf{v} \rangle + \Vert \mathbf{v} \Vert^2    \le    \Vert \mathbf{u} \Vert^2 + 2\Vert\mathbf{u} \Vert\cdot\Vert \mathbf{v} \Vert + \Vert \mathbf{v} \Vert^2$, que pode ser reescrita como $\langle \mathbf{u},\mathbf{u} \rangle + 2\langle \mathbf{u},\mathbf{v} \rangle + \langle \mathbf{v},\mathbf{v} \rangle    \le    (\Vert \mathbf{u} \Vert + \Vert \mathbf{v} \Vert)^2$, e simplificada mais ainda para $$\langle \mathbf{w},\mathbf{w} \rangle    \le    (\Vert \mathbf{u} \Vert + \Vert \mathbf{v} \Vert)^2$$            Mas, resgatando a definição da distância Euclidiana $d(\mathbf{x}, \mathbf{y})^2  = \sum_{i=1}^m (x_i - y_i)^2 = \langle \mathbf{w} , \mathbf{w} \rangle$, e como definiu-se $\mathbb{w} = \mathbb{x} - \mathbb{y}$, $\mathbb{u} = \mathbb{x} - \mathbb{z}$, $\mathbb{v} = \mathbb{z} - \mathbb{y}$, conclui-se que $$d(\mathbf{x}, \mathbf{y})    \le    d(\mathbf{x}, \mathbf{z}) + d(\mathbf{z}, \mathbf{y})$$
    $\quad\square$

## Questão 3

Uma função de dissimilaridade comum para comparar *strings* é a distância de edição. Dadas duas *strings* $x$ e $y$ de tamanhos $n$ e $m$ respectivamente, $d(x, y)$ calcula o custo mínimo, em termos de inserções, remoções e substituições, necessário para transformar $x$ em $y$.
É possível calcular $d(x, y)$ utilizando programação dinâmica a partir da recorrência descrita a seguir.

Se ambas as *strings* são vazias,
$$
d(\texttt{''}, \texttt{''}) =  0.
$$

Se a primeira *string* é vazia, mas a segunda não,
$$d(\texttt{''}, y[1, \dots, j])  =  \sum_{k = 1}^{j} \texttt{insert}(y[k]),\ \ \  1 \le j \le m.$$

Se a segunda *string* é vazia mas a primeira não,
$$d(x[1, \dots, i], \texttt{''})  =  \sum_{k = 1}^{i} \texttt{delete}(x[k]),\ \ \  1 \le i \le n.$$

Se ambas as *strings* são não vazias e $x_i  =  y_j$,
$$d(x[1,\dots, i], y[1,\dots, j]) = d(x[1,\dots, i - 1], y[1,\dots, j - 1])$$

Se ambas as *strings* são não vazias e $x_i  \neq  y_j$,
$$
\begin{align*}
	d(x[1\dots i], y[1,\dots, j]) & =  
		\text{min}
		\begin{cases}
			d(x[1,\dots, i - 1], y[1, \dots, j]) + \texttt{delete}(x[i]) \\
			d(x[1,\dots, i], y[1, \dots, j - 1]) + \texttt{insert}(y[j])\\
			d(x[1,\dots, i - 1], y[1,\dots, j - 1]) + \texttt{substitute}(x[i], y[j])
		\end{cases}
\end{align*}
$$

Na equação acima, $\texttt{''}$ é a *string* vazia e $x[1,\dots, i]$ denota a *substring* de $x$ que consiste dos primeiros $i$ símbolos de $x$.
Além disso, para um símbolo $a$, $\texttt{delete}(a)$ e $\texttt{insert}(a)$ correspondem respectivamente aos custos (positivos) de remover ou inserir o símbolo $a$.
Finalmente, para os símbolos $a$ e $b$, $\texttt{substitute}(a, b)$ corresponde ao custo (positivos) de substituir o símbolo $a$ pelo símbolo $b$.
1. Mostre, através de um exemplo de atribuição de pesos/custos para as operações disponíveis, que no caso geral $d(x, y)$ não é uma métrica.
2. Mostre que é possível atribuir custos para as funções $\texttt{delete}$, $\texttt{insert}$ e $\texttt{substitute}$ de forma que $d(x, y)$ seja uma métrica. **Obs:** você deve provar que seu caso específico é de fato uma métrica.



1. 
Para $d$ não ser uma métrica, basta que uma das 4 propriedade falhe. Se o peso atribuído para `insert` e `delete` forem diferentes, então segue que a propriedade de simetria falha. Por exemplo: 

Seja $x = \text{'a'}$ e $y = \text{''}$, $d(x,y)$ é a distância para levar x para y, nesse caso $d(x,y)=custo($`delete`$)$; e $d(y,x)$ é a distância para levar y para x, nesse caso $d(y,x)=custo($`insert`$)$. 

A propriedade de simetria pede que $d(x,y) = d(y,x)$, mas, no caso geral, é permitido $custo($`insert`$) \neq custo($`delete`$)$. As duas afirmações são incompatíveis.


2. 
Para $d$ ser uma métrica, as quatro propriedades devem ser garantidas:
- $d(x, y) \ge 0$
    A distância de edição, também conhecida como distância de Levenshtein, garante, por definição, que o custo de qualquer edição é $>0$. Como a distância é um somatório dos custos de uma sequência de edições, então $d(x,y)>0$.

    $\quad\square$
- $d(x, y) = 0 \iff x = y$
    - $x = y \Rightarrow d(x, y) = 0$: x = y é a condição de gatilho da instrução 4, que tem o efeito de truncar, recursivamente, o último caractére de ambas as strings. Mas, como as strings originais são iguais, um truncamento idêntico em ambas preserva a igualdade. O processo segue até o último caractére ser truncado, e obtém-se que $d(x[1], y[1]) = d(\texttt{''}, \texttt{''}) =  0$

    $\quad\square$
    - $d(x, y) = 0 \Rightarrow x = y$: Algoritmicamente, $d(x, y) = 0$ é atingivel a partir do caso de fronteira $d(\texttt{''}, \texttt{''}) =  0$, que por sua vez é atingível a partir do truncamento de $x[1], y[1]$ quando $x[1]=y[1]$. Este por sua vez, é atingido a partir do truncamento de $x[1\dots 2], y[1\dots 2]$ quando $x[2]=y[2]$, e assim por diante. Por indução, a única possibilidade para $d(x, y) = 0$ acontecer é quando as strings originais são iguais.

    $\quad\square$
 
- $d(x, y) = d(y, x)$
    É o caso oposto do item 1. Da mesma forma que o caso geral permite $custo($`insert`$) \neq custo($`delete`$)$, também permite o complementar $custo($`insert`$) = custo($`delete`$)$. A simetria é garantida desde que esses custos sejam iguais.

    $\quad\square$
- $d(x, y) \le d(x, z) + d(z, y)$.
    Todo `substitute` pode ser emulado por um `insert` e um `delete`. Mas, como $d$ retorna o menor custo de uma sequência de edições. Se assumirmos que o menor custo de alguma edição foi de um `substitute`, se essa edição pudesse ser substituida por um par (`insert`,`delete`) que custasse menos, então o custo desse par substituiria o custo do `substitute`. Isto é, uma contradição. Logo, pode-se afirmar
    que $custo($`substitute`$) \leq custo($`insert`$) + custo($`delete`$)$.

    Como $d(x, y)$ é dado por uma sequência de `substitute`, cada uma podendo ser emulada por um `insert` e um `delete`, em um de dois casos por etapa: (i) $x[i] \neq y[i] \rightarrow$ `insert` $y[i]$ seguido de `delete` $x[i]$;
    (ii) $x[i] \neq y[i] \rightarrow$ Insira um carcatére qualquer e depois o remova — então, a desigualdade de custos se extende para a sua soma, e é correto afirmar a desigualdade $d(x, y) \le d(x, z) + d(z, y)$.

    $\quad\square$

*Dynamic Time Warping* (DTW) é uma técnica que permite medir a "distância" entre duas séries temporais, que, intuitivamente, sejam parecidas (ou, sendo mais preciso, que "deveriam" ter uma distância "pequena"), mas que por diferenças de fase, não tenham sua distância bem representada por medidas mais simples, como a diferença em y, obtida subtraindo uma série da outra. A única restrição sobre essas séries é que suas amostras sejam equidistantes no tempo. 

Um exemplo de par de séries que precisam de uma técnica como DTW são duas séries sinusoidais de mesma frequência, mas defasadas de um ângulo substancial, por exemplo 90°. Cada meio período tem uma grande área entre as séries superior e inferior, que seria capturada por uma medida de distância simples considerando amplitude ingenuamente, e isso seria um dado espúrio.

A técnica DTW funciona encontrando, para cada amostra de uma série X, a distância a cada ponto da outra série Y, chamado *custo local*. Isso é feito para cada par de pontos (x,y), e é definida uma função de custo que soma as distâncias de cada par de pontos. A minimização dessa função encontra o *caminho de distorção ótima* para uma dada matriz de custos locais, e a distância associada ao ponto ótimo é a distância do DTW.
