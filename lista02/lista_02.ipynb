# Questão 1

Mostre que a distância de *Manhattan* (vide Aula 8) é uma métrica.

A distância de Manhattan pode ser definida como segue:
Dados $\mathbf{x} \in \mathbb{R}^m$ e $\mathbf{y} \in \mathbb{R}^m$:

$$d(\mathbf{x}, \mathbf{y}) =  \Vert \mathbf{x} - \mathbf{y} \Vert_1  = {\sum_{i=1}^m |x_i - y_i|}. $$

Uma função de dissmilaridade é dita uma métrica quando possui as seguintes propriedades:
1. $d(x, y) \ge 0$
2. $d(x, y) = 0 \iff x = y$
3. $d(x, y) = d(y, x)$
4. $d(x, y) \le d(x, z) + d(z, y)$.

### **Demonstração**
1. A distância de Manhattan é definida como uma soma de módulos. Por definição, $|x| \ge 0$, e como a soma de reais positivos sempre resulta num real positivo, então $d(\mathbf{x}, \mathbf{y}) \ge 0$ 

    $\quad\square$.

2. 
    - $d(x, y) = 0 \Rightarrow x = y$:$\quad$ $\sum_{i=1}^m |x_i - y_i| > 0$ ocorrerá sempre que $\mathbf{x} \ne \mathbf{y}$, pois, nessa condição, os componentes de $\mathbf{x}$ vão diferir de $\mathbf{y}$ em pelo menos uma coordenada, e o somando será não nulo em cada coordenada em que isto ocorrer. Mas, como demonstrado em 1, $d(\mathbf{x}, \mathbf{y}) \ge 0$. Assim, por contraposição, $d(\mathbf{x},\mathbf{y}) = 0 \Rightarrow \mathbf{x} = \mathbf{y}$. 

    - $x = y \Rightarrow d(x, y) = 0$:$\quad$ Se $\mathbf{x} = \mathbf{y}$, as coordenadas de $\mathbf{x}$ e $\mathbf{y}$ serão iguais em todas as dimensões, então $\sum_{i=1}^m |x_i - y_i| = 0$

    $\quad\square$.

3. Dentro do somando $|x_i - y_i|$, $x_i$ e $y_i$ são reais, então, pela simetria da operação de módulo $|x| = |-x|$ sobre os reais, pode-se afirmar que $|x_i - y_i| = |y_i - x_i|$. Portanto, $d(\mathbf{x}, \mathbf{y}) =  {\sum_{i=1}^m |x_i - y_i|} = {\sum_{i=1}^m |y_i - x_i|} = d(\mathbf{y}, \mathbf{x})$

    $\quad\square$.

4. $d(x, z) + d(z, y) \stackrel{\text{def}}{=} \sum_{i=1}^m |x_i - z_i| + \sum_{i=1}^m |z_i - y_i| \quad= \sum_{i=1}^m |x_i - z_i| + |z_i - y_i|$. Pela desigualdade triangular, $|u+v| \le |u| + |v|$, e faça $u = x_i - z_i$ e $v = z_i - y_i$. Logo: $|(x_i - z_i) + (z_i - y_i)| = |x_i - y_i| \;\;\le\;\; |x_i - z_i| + |z_i - y_i|$. Como a soma preserva desigualdades, pode-se fazer $\sum_{i=1}^m |x_i - y_i| \le \sum_{i=1}^m |x_i - z_i| + |z_i - y_i|$. Isto é, $d(\mathbf{x}, \mathbf{y}) \le  d(\mathbf{x}, \mathbf{z}) + d(\mathbf{z}, \mathbf{y})$. 

    $\quad\square$.

**Dica:** você pode usar a desigualdade de *Cauchy-Schwarz*.

A distância Euclidian é definida como segue:
Dados $\mathbf{x} \in \mathbb{R}^m$ e $\mathbf{y} \in \mathbb{R}^m$:

$$d(\mathbf{x}, \mathbf{y}) =  \Vert \mathbf{x} - \mathbf{y} \Vert_2  = \sqrt{\sum_{i=1}^m (x_i - y_i)^2}.$$

Uma função de dissmilaridade é dita uma métrica quando possui as seguintes propriedades:
1. $d(x, y) \ge 0$
2. $d(x, y) = 0 \iff x = y$
3. $d(x, y) = d(y, x)$
4. $d(x, y) \le d(x, z) + d(z, y)$.

### **Demonstração**
1. O contradomínio da função raíz quadrada é $\left\{ y \in \mathbb{R} \,|\, y \ge 0 \right\}$.

    $\quad\square$.

2. 
    - $d(x, y) = 0 \Rightarrow x = y$:$\quad$ $\sum_{i=1}^m (x_i - y_i)^2 > 0$ ocorrerá sempre que $\mathbf{x} \ne \mathbf{y}$, pois, nessa condição, os componentes de $\mathbf{x}$ vão diferir de $\mathbf{y}$ em pelo menos uma coordenada, e o somando será não nulo em cada coordenada em que isto ocorrer. Mas, como demonstrado em 1, $d(\mathbf{x}, \mathbf{y}) \ge 0$. Assim, por contraposição, $d(\mathbf{x},\mathbf{y}) = 0 \Rightarrow \mathbf{x} = \mathbf{y}$. 

    - $x = y \Rightarrow d(x, y) = 0$:$\quad$ Se $\mathbf{x} = \mathbf{y}$, as coordenadas de $\mathbf{x}$ e $\mathbf{y}$ serão iguais em todas as dimensões, então $\sum_{i=1}^m (x_i - y_i) = 0$

    $\quad\square$.

3. Dentro do somando $(x_i - y_i)^2$, $x_i$ e $y_i$ são reais, então, pela simetria da operação $x^2 = (-x)^2$ sobre os reais, pode-se afirmar que $(x_i - y_i) = (y_i - x_i)$. Portanto, $d(\mathbf{x}, \mathbf{y}) =  \sqrt{\sum_{i=1}^m (x_i - y_i)^2} = \sqrt{\sum_{i=1}^m (y_i - x_i)^2} = d(\mathbf{y}, \mathbf{x})$

    $\quad\square$.

## Questão 3

Uma função de dissimilaridade comum para comparar *strings* é a distância de edição. Dadas duas *strings* $x$ e $y$ de tamanhos $n$ e $m$ respectivamente, $d(x, y)$ calcula o custo mínimo, em termos de inserções, remoções e substituições, necessário para transformar $x$ em $y$.
É possível calcular $d(x, y)$ utilizando programação dinâmica a partir da recorrência descrita a seguir.

Se ambas as *strings* são vazias,
$$
d(\texttt{''}, \texttt{''}) =  0.
$$

Se a primeira *string* é vazia, mas a segunda não,
$$d(\texttt{''}, y[1, \dots, j])  =  \sum_{k = 1}^{j} \texttt{insert}(y[k]),\ \ \  1 \le j \le m.$$

Se a segunda *string* é vazia mas a primeira não,
$$d(x[1, \dots, i], \texttt{''})  =  \sum_{k = 1}^{i} \texttt{delete}(x[k]),\ \ \  1 \le i \le n.$$

Se ambas as *strings* são não vazias e $x_i  =  y_j$,
$$d(x[1,\dots, i], y[1,\dots, j]) = d(x[1,\dots, i - 1], y[1,\dots, j - 1])$$

Se ambas as *strings* são não vazias e $x_i  \neq  y_j$,
$$
\begin{align*}
	d(x[1\dots i], y[1,\dots, j]) & =  
		\text{min}
		\begin{cases}
			d(x[1,\dots, i - 1], y[1, \dots, j]) + \texttt{delete}(x[i]) \\
			d(x[1,\dots, i], y[1, \dots, j - 1]) + \texttt{insert}(y[j])\\
			d(x[1,\dots, i - 1], y[1,\dots, j - 1]) + \texttt{substitute}(x[i], y[j])
		\end{cases}
\end{align*}
$$

Na equação acima, $\texttt{''}$ é a *string* vazia e $x[1,\dots, i]$ denota a *substring* de $x$ que consiste dos primeiros $i$ símbolos de $x$.
Além disso, para um símbolo $a$, $\texttt{delete}(a)$ e $\texttt{insert}(a)$ correspondem respectivamente aos custos (positivos) de remover ou inserir o símbolo $a$.
Finalmente, para os símbolos $a$ e $b$, $\texttt{substitute}(a, b)$ corresponde ao custo (positivos) de substituir o símbolo $a$ pelo símbolo $b$.
1. Mostre, através de um exemplo de atribuição de pesos/custos para as operações disponíveis, que no caso geral $d(x, y)$ não é uma métrica.
2. Mostre que é possível atribuir custos para as funções $\texttt{delete}$, $\texttt{insert}$ e $\texttt{substitute}$ de forma que $d(x, y)$ seja uma métrica. **Obs:** você deve provar que seu caso específico é de fato uma métrica.
